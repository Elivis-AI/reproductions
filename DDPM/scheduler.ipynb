```python
from ddpm import *

n_steps = 1000
constant = DDPM(constant_beta_schedule(n_steps)).p
linear = DDPM(linear_schedule(n_steps)).p
sigmoid = DDPM(sigmoid_schedule(n_steps)).p
cosine = DDPM(cosine_schedule(n_steps)).p
sqrt = DDPM(sqrt_schedule(n_steps)).p
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, n_steps, n_steps)

plt.figure(figsize=(8, 5))
plt.plot(x, constant, label='constant')
plt.plot(x, linear, label='linear')
plt.plot(x, sigmoid, label='sigmoid')
plt.plot(x, cosine, label='cosine')
plt.plot(x, sqrt, label='sqrt')

plt.xlabel('steps($t$)')
plt.ylabel('$\\sqrt{\\gamma(t)} = \\sqrt{\\bar{\\alpha}_t}$')
plt.legend()
plt.show()
```
